# 4. Data Cleaning

## Check Missing Values

First inspecting the missing values before making any changes.

```python
missing_values = df.isnull().sum()
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage.round(2)
}).sort_values("Missing Values", ascending=False)

display(missing_summary[missing_summary["Missing Values"] > 0])
```

## Handle Missing Values

Based on the dataset,requirements and the project questions:

- **`children`**: missing values are treated as 0 because it is required for `Total_Guests`.
- **`country`**: missing values are retained as `"Unknown"` because country is required later for the top-5 completed-bookings analysis.
- **`agent`**: retained with missing values because it is not needed for the required business questions and there is no justified value to impute.
- **`company`**: removed because it has a very large number of missing values and is not required for the specified business questions.

No artificial identifier values are inserted into **`agent`** or **`company`**.

```python
# Handles 'children'
df["children"] = df["children"].fillna(0)

# Handle 'country'
df["country"] = df["country"].fillna("Unknown")

# Remove 'company' because of extensive missingness and lack of relevance
# to the assessment's required questions.
df.drop(columns=["company"], inplace=True)

print("Missing values after initial handling:")
display(df.isnull().sum().sort_values(ascending=False).head(10))
```

## Checking Duplicate Rows

```python
duplicate_count = df.duplicated().sum()
print("Duplicate rows:", duplicate_count)
```

```python
# Remove exact duplicate rows
df = df.drop_duplicates().reset_index(drop=True)

print("Dataset shape after duplicate removal:", df.shape)
print("Remaining duplicate rows:", df.duplicated().sum())
```

## Correcting Data Types

For this dataset:
- `reservation_status_date` is a genuine date field and should be `datetime64[ns]`.
- `children` can be converted to integer after missing values are handled.
- The arrival year/month/week/day columns are **date components**, so they should not be converted individually to datetime.

The four arrival components are kept in their original appropriate types.

```python
# Converting the genuine date column
df["reservation_status_date"] = pd.to_datetime(
    df["reservation_status_date"],
    errors="coerce"
)

# Converting children to integer after filling missing values
df["children"] = df["children"].astype(int)

print(df[[
    "children",
    "reservation_status_date"
]].dtypes)
```

## Create a Complete Arrival Date

The dataset stores arrival date information in separate columns:
- `arrival_date_year`
- `arrival_date_month`
- `arrival_date_day_of_month`

Creating one additional `Arrival_Date` column for easier time-based analysis. Keep the original component columns as well.

```python
df["Arrival_Date"] = pd.to_datetime(
    df["arrival_date_year"].astype(str)
    + "-"
    + df["arrival_date_month"]
    + "-"
    + df["arrival_date_day_of_month"].astype(str),
    format="%Y-%B-%d",
    errors="coerce"
)

display(df[[
    "arrival_date_year",
    "arrival_date_month",
    "arrival_date_day_of_month",
    "Arrival_Date"
]].head())
```

## Check Invalid or Unusual Values

- Zero guests
- Zero-night stays
- Negative ADR
- Negative lead time
- Invalid cancellation indicator
- Invalid hotel values

```python
print("Zero-guest records:", (df["adults"] + df["children"] + df["babies"] == 0).sum())
print("Zero-night records:",
      (df["stays_in_weekend_nights"] + df["stays_in_week_nights"] == 0).sum())
print("Negative ADR records:", (df["adr"] < 0).sum())
print("Negative lead-time records:", (df["lead_time"] < 0).sum())

print("\nUnexpected is_canceled values:",
      sorted(set(df["is_canceled"].dropna().unique()) - {0, 1}))

print("\nHotel values:")
print(df["hotel"].value_counts(dropna=False))
```

### Treatment of unusual values

For the cleaning stage:
- Negative ADR values are invalid and are removed.
- Negative lead times are invalid and are removed.
- Records with zero guests are treated as invalid booking records and are removed.
- Zero-night stays are **not automatically deleted** because they can represent records that need business interpretation; they are flagged and retained unless the project decides otherwise.


```python
# Removes clearly invalid numeric records
invalid_adr = df["adr"] < 0
invalid_lead_time = df["lead_time"] < 0
zero_guests = (
    df["adults"] +
    df["children"] +
    df["babies"]
) == 0

print("Rows removed for negative ADR:", invalid_adr.sum())
print("Rows removed for negative lead time:", invalid_lead_time.sum())
print("Rows removed for zero guests:", zero_guests.sum())

df = df.loc[
    ~(invalid_adr | invalid_lead_time | zero_guests)
].copy()

df.reset_index(drop=True, inplace=True)
```

# 7. Required Basic Analysis

The assessment requires these seven metrics:
1. Total bookings
2. Total cancelled bookings
3. Cancellation rate
4. Average Daily Rate (ADR)
5. Average stay duration
6. City Hotel bookings
7. Resort Hotel bookings

```python
# Total bookings
total_bookings = len(df)

# Total cancelled bookings
total_cancelled = (
    df["Booking_Status"] == "Cancelled"
).sum()

# Cancellation rate
cancellation_rate = (
    total_cancelled / total_bookings
) * 100

# Average Daily Rate
average_adr = df["adr"].mean()

# Average stay duration
average_stay = df["Stay_Nights"].mean()

# City Hotel bookings
city_hotel_bookings = (
    df["hotel"] == "City Hotel"
).sum()

# Resort Hotel bookings
resort_hotel_bookings = (
    df["hotel"] == "Resort Hotel"
).sum()

summary_df = pd.DataFrame({
    "Metric": [
        "Total Bookings",
        "Total Cancelled Bookings",
        "Cancellation Rate (%)",
        "Average ADR",
        "Average Stay Duration (Nights)",
        "City Hotel Bookings",
        "Resort Hotel Bookings"
    ],
    "Value": [
        total_bookings,
        total_cancelled,
        round(cancellation_rate, 2),
        round(average_adr, 2),
        round(average_stay, 2),
        city_hotel_bookings,
        resort_hotel_bookings
    ]
})

display(summary_df)
```

## Additional Descriptive Analysis

## Cancellation Rate by Hotel

This is useful for the business question: which hotel has the higher cancellation percentage?

```python
hotel_cancellation = (
    df.groupby("hotel")
      .agg(
          Total_Bookings=("hotel", "size"),
          Cancelled_Bookings=("Booking_Status",
                             lambda x: (x == "Cancelled").sum())
      )
)

hotel_cancellation["Cancellation_Rate_%"] = (
    hotel_cancellation["Cancelled_Bookings"]
    / hotel_cancellation["Total_Bookings"]
    * 100
)

hotel_cancellation = hotel_cancellation.sort_values(
    "Cancellation_Rate_%",
    ascending=False
)

display(hotel_cancellation)
```

## Monthly Booking Summary

Displaying using `Arrival_Date` to create a month-level summary.

```python
monthly_bookings = (
    df.groupby(
        df["Arrival_Date"].dt.month_name()
    )
    .size()
    .sort_values(ascending=False)
)

display(monthly_bookings.to_frame("Total Bookings"))
```

## Lead Time Category

cancellation analysis:
- 0 to 30 days
- 31 to 90 days
- 91 to 180 days
- More than 180 days

```python
bins = [0, 30, 90, 180, np.inf]
labels = [
    "0-30 Days",
    "31-90 Days",
    "91-180 Days",
    "180+ Days"
]

df["Lead_Time_Category"] = pd.cut(
    df["lead_time"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

lead_time_analysis = (
    df.groupby("Lead_Time_Category", observed=False)
      .agg(
          Total_Bookings=("Booking_Status", "size"),
          Cancelled_Bookings=(
              "Booking_Status",
              lambda x: (x == "Cancelled").sum()
          )
      )
)

lead_time_analysis["Cancellation_Rate_%"] = (
    lead_time_analysis["Cancelled_Bookings"]
    / lead_time_analysis["Total_Bookings"]
    * 100
)

display(lead_time_analysis)
```

# BeautifulSoup Web Scraping

## Objective

To Collect approximately 10 country-level tourism records from a public
tourism/informational website using requests and BeautifulSoup.

The scraped data will contain:
- Country
- Tourist Arrivals

The final dataset will be saved as scraped_data.csv

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

### Defining the webpage

```python
url = "https://www.pib.gov.in/PressReleasePage.aspx?PRID=2153606&lang=2&reg=48"

response = requests.get(url)

print("Status Code:", response.status_code)
```

If you get 200, continue.

### Creating the BeautifulSoup object

```python
soup = BeautifulSoup(response.text, "html.parser")

print("BeautifulSoup object created successfully.")
print("Page title:", soup.title.get_text(strip=True))
```

### Inspecting the webpage, Before Scraping

```python
tables = soup.find_all("table")
print("Number of tables found:", len(tables))
```

```python
for i, table in enumerate(tables):
    print(f"\nTABLE {i}")
    print(table.get_text(" | ", strip=True)[:1000])
    print("-" * 80)
```

## Finding the tourism table

```python
tourism_table = tables[1]

print("Tourism table selected successfully.")
```

```python
# Get all rows in the table
rows = tourism_table.find_all("tr")

print("Number of rows:", len(rows))
```

```python
# Inspecting first 5 rows
for row in rows[:5]:
    cells = row.find_all(["th", "td"])
    print([cell.get_text(" ", strip=True) for cell in cells])
```

### Extracting the data

```python
tourism_records = []

for row in rows:
    cells = row.find_all("td")
    
    if len(cells) >= 6:
        serial_no = cells[0].get_text(" ", strip=True)
        country = cells[1].get_text(" ", strip=True)
        arrivals_2019 = cells[2].get_text(" ", strip=True)
        arrivals_2022 = cells[3].get_text(" ", strip=True)
        arrivals_2023 = cells[4].get_text(" ", strip=True)
        arrivals_2024 = cells[5].get_text(" ", strip=True)
        
        tourism_records.append({
            "Country": country,
            "FTA_2019": arrivals_2019,
            "FTA_2022": arrivals_2022,
            "FTA_2023": arrivals_2023,
            "FTA_2024": arrivals_2024
        })

print("Records collected:", len(tourism_records))
```

### Creating Data Frame

```python
tourism_df = pd.DataFrame(tourism_records)

tourism_df.head(15)
```

### Keeping 10 records

```python
tourism_df = tourism_df.head(10).copy()

tourism_df
```

### Convert arrival values to numbers

Currently they contain commas:
*18,04,586*

Needed:
*1804586*

```python
arrival_columns = [
    "FTA_2019",
    "FTA_2022",
    "FTA_2023",
    "FTA_2024"
]

for column in arrival_columns:
    tourism_df[column] = (
        tourism_df[column]
        .str.replace(",", "", regex=False)
        .str.strip()
    )
    
    tourism_df[column] = pd.to_numeric(
        tourism_df[column],
        errors="coerce"
    )
```

```python
# ============================================================
# BONUS ANALYSIS: HOTEL BOOKINGS vs TOURISM ARRIVALS
# ============================================================

import pandas as pd

# Load datasets
hotel_df = pd.read_csv("cleaned_hotel_bookings.csv")
tourism_df = pd.read_csv("scraped_data.csv")

# Map hotel dataset country codes to tourism country names
country_mapping = {
    "USA": "United States",
    "GBR": "United Kingdom",
    "AUS": "Australia",
    "CAN": "Canada",
    "MYS": "Malaysia",
    "LKA": "Sri Lanka",
    "DEU": "Germany",
    "FRA": "France",
    "BGD": "Bangladesh",
    "IND": "India",
    "ESP": "Spain",
    "ITA": "Italy",
    "NLD": "Netherlands",
    "IRL": "Ireland",
    "JPN": "Japan",
    "RUS": "Russian Fed.",
    "THA": "Thailand",
    "SGP": "Singapore",
    "KOR": "Rep. of Korea",
    "CHE": "Switzerland",
    "SWE": "Sweden",
    "NOR": "Norway",
    "DNK": "Denmark",
    "BEL": "Belgium",
    "AUT": "Austria",
    "PRT": "Portugal",
    "POL": "Poland",
    "BRA": "Brazil",
    "ARG": "Argentina",
    "CHN": "China",
    "HKG": "Hong Kong",
    "ISR": "Israel",
    "ZAF": "South Africa",
    "NZL": "New Zealand"
}

# Create readable country names
hotel_df["Country_Name"] = hotel_df["country"].map(country_mapping)

# Count hotel bookings by country-
hotel_bookings_country = (
    hotel_df
    .groupby("Country_Name", dropna=True)
    .size()
    .reset_index(name="Hotel_Bookings")
)

# Merge with tourism data
bonus_df = pd.merge(
    hotel_bookings_country,
    tourism_df,
    left_on="Country_Name",
    right_on="Country",
    how="inner"
)

# Calculate booking share and comparison metric
bonus_df["Hotel_Booking_Rank"] = (
    bonus_df["Hotel_Bookings"]
    .rank(method="min", ascending=False)
    .astype(int)
)

bonus_df["Tourism_Arrival_Rank_2024"] = (
    bonus_df["FTA_2024"]
    .rank(method="min", ascending=False)
    .astype(int)
)

# Difference between tourism rank and hotel booking rank
bonus_df["Rank_Difference"] = (
    bonus_df["Tourism_Arrival_Rank_2024"]
    - bonus_df["Hotel_Booking_Rank"]
)

# Sort by hotel bookings
bonus_df = bonus_df.sort_values(
    "Hotel_Bookings",
    ascending=False
).reset_index(drop=True)

# Display result
print("=" * 70)
print("BONUS ANALYSIS: HOTEL BOOKINGS vs TOURISM ARRIVALS")
print("=" * 70)

display(
    bonus_df[
        [
            "Country_Name",
            "Hotel_Bookings",
            "FTA_2019",
            "FTA_2022",
            "FTA_2023",
            "FTA_2024",
            "Hotel_Booking_Rank",
            "Tourism_Arrival_Rank_2024",
            "Rank_Difference"
        ]
    ]
)

# Save bonus analysis
bonus_df.to_csv(
    "hotel_bookings_vs_tourism_bonus.csv",
    index=False
)

print("\nBonus analysis saved successfully as:")
print("hotel_bookings_vs_tourism_bonus.csv")
print("\nCountries successfully matched:", len(bonus_df))
```
